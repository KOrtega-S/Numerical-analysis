# Problem 2 — von Neumann Stability of the Explicit Heat Equation

$$\frac{\partial u}{\partial t}=\kappa\,\frac{\partial^{2}u}{\partial x^{2}},\quad \kappa>0,\qquad \frac{u_j^{n+1}-u_j^{n}}{\Delta t}=\kappa\,\frac{u_{j+1}^{n}-2u_j^{n}+u_{j-1}^{n}}{\Delta x^{2}}$$

## Step 1. Explicit form

$$r=\frac{\kappa\,\Delta t}{\Delta x^{2}}>0$$

$$u_j^{n+1}=u_j^{n}+r\left(u_{j+1}^{n}-2u_j^{n}+u_{j-1}^{n}\right)$$

## Step 2. Error equation

The scheme is linear, so the error $e_j^n=\tilde{u}_j^n-u_j^n$ satisfies the same equation:

$$e_j^{n+1}=e_j^{n}+r\left(e_{j+1}^{n}-2e_j^{n}+e_{j-1}^{n}\right)$$

Any perturbation is a sum of Fourier modes that evolve independently, so one mode is enough.

## Step 3. Fourier mode

$$e_j^{n}=G^{\,n}e^{\,ikj\Delta x}=G^{\,n}e^{\,ij\theta},\qquad \theta=k\Delta x\in[-\pi,\pi]$$

$$e_{j\pm1}^{n}=G^{\,n}e^{\,ij\theta}e^{\pm i\theta},\qquad e_j^{n+1}=G^{\,n}e^{\,ij\theta}\,G$$

## Step 4. Amplification factor

Substitute and divide by $G^{\,n}e^{\,ij\theta}\neq0$:

$$G=1+r\left(e^{\,i\theta}-2+e^{-i\theta}\right)$$

Using $e^{\,i\theta}+e^{-i\theta}=2\cos\theta$:

$$G=1-2r\left(1-\cos\theta\right)$$

Using $1-\cos\theta=2\sin^{2}(\theta/2)$:

$$\boxed{G(\theta)=1-4r\sin^{2}\!\left(\frac{\theta}{2}\right)}$$

$G$ is real, so $|G|\le1 \iff -1\le G\le 1$.

## Step 5. Condition $|G|\le 1$ for all $\theta$

Let $s=\sin^{2}(\theta/2)\in[0,1]$:

$$-1\le 1-4rs\le 1$$

**Right:** $1-4rs\le1 \iff rs\ge0$ — always true.

**Left:**

$$-1\le1-4rs\iff rs\le\frac12$$

Worst case $s_{\max}=1$ at $\theta=\pm\pi$:

$$r\le\frac12$$

## Step 6. Necessity

If $r>\tfrac12$, the mode $\theta=\pi$, i.e. $e_j^n=(1-4r)^n(-1)^j$, has

$$G(\pi)=1-4r<-1\ \Rightarrow\ |G|^n\to\infty$$

so the scheme is unstable.

## Step 7. Result

$$\boxed{\frac{\kappa\,\Delta t}{\Delta x^{2}}\le\frac12\iff\Delta t\le\frac{\Delta x^{2}}{2\kappa}}$$

The scheme is conditionally stable.
